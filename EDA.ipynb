```python
import pandas as pd
import re
import numpy as np


df = pd.read_csv('books.csv', encoding="utf-8-sig")


df.columns = df.columns.str.strip().str.lower()


df.rename(columns={'disponibilité': 'disponibilite'}, inplace=True)


def nettoyer_description(desc):
    if pd.isna(desc):
        return ""
    desc = re.sub(r'\s+', ' ', desc)
    desc = re.sub(r'[^A-Za-z0-9À-ÖØ-öø-ÿ\s]', '', desc)  
    return desc.strip()

df['description'] = df['description'].apply(nettoyer_description)


df['prix'] = df['prix'].replace('[\£,]', '', regex=True).astype(float)

df['disponibilite'] = df['disponibilite'].str.extract(r'(\d+)').astype(float).fillna(0).astype(int)

note_mapping = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5}
df['note'] = df['note'].map(note_mapping)


df['description'] = df['description'].replace('', np.nan)
df['description'] = df['description'].fillna(df['titre'])

df.to_csv('livres_bruts.csv', index=False)

data = pd.read_csv('livres_bruts.csv')

data
print(df["note"].dtype)


```

```python
import pandas as pd
from langdetect import detect
from stop_words import get_stop_words



data["lang"] = data["description"].apply(lambda x: detect(str(x)) if pd.notnull(x) else "unknown")


stop_fr = set(get_stop_words("french"))
stop_en = set(get_stop_words("english"))

stop_all = list(set(stop_en).union(set(stop_fr)))


print(data["lang"].value_counts())



```

```python
data
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction import text
from sklearn.metrics.pairwise import cosine_similarity
import joblib


vectorizer = TfidfVectorizer(stop_words=stop_all)

matrix = vectorizer.fit_transform(data['description'])

similarity = cosine_similarity(matrix, matrix)


joblib.dump(similarity, "similarite_livres.pkl")
joblib.dump(vectorizer, "vectoriseur.pkl")
joblib.dump(data, "data_livres.pkl")


def recommander_multilingue(titre, top_n=5):
    if titre not in data['titre'].values:
        return []
    
    idx = data.index[data['titre'] == titre][0]
    scores = list(enumerate(similarity[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    scores = [s for s in scores if s[0] != idx][:top_n]
    
    livres_suggest = data.iloc[[s[0] for s in scores]]
    return livres_suggest[['titre', 'prix', 'note', 'disponibilite']].to_dict(orient='records')

exemple = recommander_multilingue("Titre d'exemple", 3)
print(exemple)









```
